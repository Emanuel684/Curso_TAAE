```python
import pandas as pd
matriculas=pd.read_excel('https://robertohincapie.com/data/libro_merge.xlsx', sheet_name='matriculas')
usuarios=pd.read_excel('https://robertohincapie.com/data/libro_merge.xlsx', sheet_name='usuarios')
cursos=pd.read_excel('https://robertohincapie.com/data/libro_merge.xlsx', sheet_name='cursos')

print('Cursos:\n', cursos)
print('\nUsuarios:\n', usuarios)
print('\nMatriculas:\n',matriculas)
```

```python
#Agreguemos el nombre de los usuarios al dataframe actual y creemos otro dataframe
resumen=matriculas.merge(right=usuarios, how='left', left_on='IDEstudiante (*)', right_on='ID (*)')
resumen
```

```python
#Reorganicemos un poco el resultado, cambiando el nombre de las columnas que encontramos y
#eliminando el campo de ID que queda sobrando
resumen=resumen.rename(columns={'Nombres':'Nombres estudiante','Apellidos':'Apellidos estudiante'})
resumen=resumen.drop('ID (*)', axis=1)
resumen.head()
```

```python
#Ahora traigamos el nombre del profesor y el curso al dataframe
resumen=resumen.merge(right=cursos, how='left', left_on='Curso (*)', right_on='Curso (*)')
resumen=resumen.rename(columns={'Nombres':'Nombre curso'})
resumen.head()
```

```python
#Finalmente agreguemos el nombre del profesor a DF actual
resumen=resumen.merge(right=usuarios, how='left', left_on='IDProfesor', right_on='ID (*)')
resumen=resumen.rename(columns={'Nombres':'Nombres Profesor','Apellidos':'Apellidos Profesor'})
resumen=resumen.drop('ID (*)', axis=1)
resumen
```
