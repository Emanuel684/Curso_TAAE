```python
#Vamos a calcular una nueva columna que tiene el precio por metro cuadrado
data['Valor/m2'] = data['Precio']/data['Area_Construida']
data
```

```python
#Convirtamos la columna de antigüedad en un número, para poder graficar el valor por metro cuadrado con respecto a la antigüedad de la propiedad.
data['Antiguedad'].unique()
```

```python
data['Antiguedad']=data['Antiguedad'].replace({'Entre 0 y 5 años':2.5, 'Remodelado':20, 'Entre 10 y 20 años':15,
       'Entre 5 y 10 años':7.5, 'Más de 20 años':30})

plt.plot(data['Antiguedad'], data['Valor/m2'],'.')
plt.xlabel('Antigüedad')
plt.ylabel('Valor por metro cuadrado')
plt.grid()
```

```python
observacion=[]
for nb in data['Baños'].values: #Esto me permite iterar sobre cada valor del número de baños.
  if(nb<=2):
    observacion.append('Pocos baños')
  else:
    observacion.append('Ninguna')
data['Observaciones']=observacion
data.head(10)
data.to_excel('Excel con observaciones.xlsx')
```

```python
def observaciones(x):
  cad=''
  if(x['Baños']>2):
    cad='Buen número de baños'
  else:
    cad='Pocos baños'
  if(x['Parqueaderos']>1):
    cad=cad+', Más de un parqueadero'
  else:
    cad=cad+', solo un parqueadero'

  return cad

data['Observaciones']=data.apply(lambda x: observaciones(x), axis=1)
data.head(10)
data.to_excel('Excel con observaciones.xlsx')
```
