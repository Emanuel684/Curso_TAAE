Vamos a comenzar importando las dos librerías. Con Matplotlib trabajaremos con un paquete llamado pyplot. Qué representa el nombre plt? Es la identificación con la que vamos a identificar todo lo que tenga que ver con pyplot

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.linspace(0,4*np.pi, 200)
y=np.sin(x)
plt.plot(x,y)
```

Este ejemplo simple involucra dos vectores: x y y. El primero para las coordenadas horizontales y el segundo para las coordenadas verticales. La curva tiene muchos más aspectos que pueden mejorarse a partir de otros parámetros.

Por ejemplo, el tipo de línea, los marcadores y el color se controla por una cadena que tiene dicha información así: https://matplotlib.org/2.1.2/api/_as_gen/matplotlib.pyplot.plot.html

'-ko' significa una linea continua, negra con marcador de círculo
':sg' representa una línea punteada, color verde y marcador de cuadrado.

Miremos estos ejemplos

```python
#x=np.linspace(0,4*np.pi, 100)
x=np.arange(0,4*np.pi, 0.5)
y1=np.cos(x)
y2=np.sin(x)
y3=np.cos(x)*np.exp(-x/3)
plt.plot(x,y1, 'x-r', label='cos(x)') #El label es para darle un nombre a la curva en un comando siguiente.
plt.plot(x,y2, 'b-', label='sin(x)')
plt.plot(x,y3, '^:g', label='curva 3')

plt.legend() #Esto crea la etiqueta de las curvas
plt.grid() #Esto activa una cuadrícula
plt.xlabel('Coordenada x')
plt.ylabel('Coordenada y')
plt.title('Curvas en función de la x')
```

Lo primero que se requiere con Matplotlib es crear figuras. Estas se crean especificando su tamaño.

```python
plt.figure(figsize=(1,1))
plt.plot(x,y1, 'x-r', label='cos(x)')
plt.figure(figsize=(12,4))
plt.plot(x,y1, 'x-r', label='cos(x)', linewidth=10)
plt.xlim(0,12)
plt.legend()
```

A veces se requiere poder colocar una figura al lado de otra. Es como crear una figura con diferentes subfiguras. Esto se logra con el comando subplot, el cuál requiere como argumento el número de filas, el número de columnas y el indicador de la figura sobre la que se está dibujando.

```python

plt.figure(figsize=(16,4))
plt.subplot(1,4,1) #1 fila y 4 columnas
plt.plot(x,y1,'o-.r')
plt.subplot(1,4,2)
plt.plot(x,y2)
plt.subplot(1,4,3)
plt.plot(x,y3)
plt.subplot(1,4,4)
plt.plot(x,y3*x)
plt.minorticks_on()
```

```python
plt.figure(figsize=(6,4))
plt.subplot(2,2,1) #1 fila y 4 columnas
plt.plot(x,y1,'o-.r')
plt.subplot(2,2,2)
plt.plot(x,y2)
plt.subplot(2,2,3)
plt.plot(x,y3)
plt.subplot(2,2,4)
plt.plot(x,y3*x)
plt.minorticks_on()
```
