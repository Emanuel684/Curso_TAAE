Vamos a realizar un proceso de filtrado de la información. Primero leeremos los datos y luego haremos un filtrado de los datos de acuerdo con cierta condición.

```python
import pandas as pd
import matplotlib.pyplot as plt

data=pd.read_excel('https://www.robertohincapie.com/data/datos_viviendas1.xlsx')
data.head()
```

```python
#Vamos a graficar alguna información del dataframe anterior, por ejemplo el precio del apartamento
data['Precio'].plot()
```

```python
#Podemos graficar también el precio con respecto al área del inmueble
plt.plot(data['Area_Construida'], data['Precio'], 'r.')
plt.xlabel('Área construida')
plt.ylabel('Precio')
plt.grid()
```

Vamos a hacer un proceso de filtrado. Tomaremos los datos uno a uno y verificaremos el precio. Si cumple con la condición, lo agregaremos a una lista y luego armaremos de nuevo un DataFrame.

El proceso de filtrado se puede realizar por medio de una expresión propia de Pandas, la cual permite definir la condición por medio de la cuál se hace el filtrado. Es una manera de resumir la operación manual anterior, dentro del modelo de DataFrame.

image.png

```python
index=data['Precio']<=500000000
index
```

```python
df3=data[index]
df3
```

```python
df3=data[      data['Precio']<=500000000          ] #instrucción de filtrado
df3.head()
```

Estas operaciones de filtrado se pueden construir de una manera más compleja. Tan complejas como pueden ser las operaciones lógicas. Si se comprendió esto anteriormente, se puede intuir cómo construirlas fácilmente.

Lo único es comprender que como se opera a nivel de arreglos, se utilizan otros operadores: and equivale a &, or equivale a |.

Por ejemplo, podemos filtrar los apartamentos que tienen un precio menor o igual a 500M, mayor o igual a 400M y con un área mayor a 400m2.

```python
df3=data[ (data['Precio']<=500000000) & (data['Area_Construida']>=100) & (data['Precio']>=400000000) ]
df3
```
