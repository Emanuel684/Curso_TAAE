Vamos a realizar diferentes procesos de reducción sobre los datos de las viviendas que hemos venido trabajando.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

data=pd.read_excel('https://www.robertohincapie.com/data/datos_viviendas1.xlsx')
data.head()
```

```python
data['Precio']
```

```python
#Vamos a calcular algunos resultados de reducción de ciertas columnas del dataframe
print('Precio promedio: ',data['Precio'].mean()) #Acá utilizamos la función MEAN de pandas.
print('Precio promedio: ',np.mean(data['Precio'])) #Acá utilizamos la función MEAN de Numpy.

print('Precio máximo: ',data['Precio'].max())
print('Precio mínimo: ',data['Precio'].min())
print('Tercer cuartil, percentil del 75%', np.percentile(data['Precio'], 75))
```
