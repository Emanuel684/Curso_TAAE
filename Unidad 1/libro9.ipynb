```python
import pandas as pd
import matplotlib.pyplot as plt
#leemos una archivo que contiene los datos de una curva en función del tiempo
curva1=pd.read_excel('https://robertohincapie.com/data/libro_join.xlsx', sheet_name='Curva 1')
#Hacemos que la columna de tiempo sea el índice del DataFrame.
curva1.index=curva1.tiempo
curva1=curva1.drop('tiempo', axis=1)

#leemos una archivo que contiene los datos de otra curva en función del tiempo
curva2=pd.read_excel('https://robertohincapie.com/data/libro_join.xlsx', sheet_name='Curva 2')
#Hacemos que la columna de tiempo sea el índice del DataFrame.
curva2.index=curva2.tiempo
curva2=curva2.drop('tiempo', axis=1)

curva2.head()
```

```python
curva2.head()
```

```python
#Cómo los índices de ambas curvas son el valor del tiempo y son iguales, podemos unirlas utilizando el método JOIN
#Unión de los dos dataframes

completo=curva1.join(curva2)

completo.plot()
plt.grid()
plt.xlabel('Tiempo')
completo.head()
```
