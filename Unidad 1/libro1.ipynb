Pandas permite cargar la información dentro de una estructura llamada un dataframe. Este es una tabla organizada en forma filas compuestas por registros y columnas que representan datos.



```python
import pandas as pd   #De ahora en adelante, las funcionalidades de pandas se ejecutarán como pd.algo
data = pd.read_excel('https://www.robertohincapie.com/data/datos_viviendas1.xlsx')
#En este momento solamente hemos cargado los datos a memoria.
```

El concepto del dataframe es el del mismo arreglo de dos dimensiones, pero especificando para cada dato, tanto su fila como su columna. De esta manera, se tiene una visión mucho más concreta de los datos, más allá de la abstracción del arreglo con filas y columnas

```python
print('Columnas del dataframe: ', data.columns)
```

```python
#Podemos buscar ciertas filas específicas del DataFrame por medio de la Expresión iloc e indicando las filas que se desea recuperar.
data.iloc[[30,10,50]] #Observe que el corchete externo es para llamar la función. El corchete interno es para indicar que no es una fila sino varias.
```

```python
#Para acceder a una fila o un conjunto de filas del dataframe, simplemente podemos utilizar el operador rango, dentro de la instrucción iloc del dataframe
#El rango es un operador que especifica el origen:final, siendo el final no incluido en el rango.
#El comando iloc del dataframe permite acceder por el número de la fila
print('Filas desde la 5 hasta la 8:\n')
data.iloc[85:]
```

```python
#Pero podemos leer también una columna específica:
print('La columna del estrato es:\n')
data[["Estrato", 'Barrio']]
```

```python
#Pero podemos combinar la columna con ciertas filas únicamente.
print('La columna del estrato para ciertas filas es:\n')
data.iloc[5:9]['Estrato']
```

```python
#Pero como usted se debe haber podido imaginar, es posible combinar las filas y las columnas en una consulta, por ejemplo:
print('La columna de precio de las primeras 10 filas es: \n')
data.iloc[:10]['Precio']
```

```python
data.iloc[30:][  ['Precio','Estrato']  ]
```

Los dataframes también pueden accederse por medio de un índice que es una marca única que representa cada fila. Para convertir alguno de los datos como índice, es expresar la columna que utilizaremos como tal. En este caso, el código del inmueble será especificado como la clave. Seguramente es el indicador único para cada apartamento o propiedad



```python
data.set_index('Cod_Inmueble', drop=True, inplace=True)
data.head()
```

Cuando tenemos un índice, podemos seleccionar datos por medio del índice. En este caso, se utiliza el comando loc[indice]



```python
data.loc[['11318-M3021864','11318-M3058418']]
#data.iloc[10]
```

Existen otras funciones muy importantes que con el tiempo estudiaremos. Entre ellas, veremos la función unique(), que retorna los valores únicos de una determinada columna. Y la función value_counts() que hace lo mismo, pero indica cuántos datos de cada uno de los valores únicos hay.

```python
data['Habitaciones'].unique()
```

```python
data['Habitaciones'].value_counts()
```

```python
data['Habitaciones'].nunique()
```
