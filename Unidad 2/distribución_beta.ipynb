La distribución Beta es una distribución de probabilidad continua definida sobre el intervalo [0,1], por lo que resulta especialmente adecuada para representar la incertidumbre acerca de una probabilidad desconocida. En lugar de asumir que el valor de una probabilidad p es fijo y conocido, la inferencia bayesiana considera que dicho valor puede tomar distintos valores entre 0 y 1, cada uno con una determinada plausibilidad.

La forma de la distribución está determinada por dos parámetros, tradicionalmente denominados alpha y beta. Estos parámetros pueden interpretarse como la cantidad de evidencia previa a favor y en contra del evento de interés. Intuitivamente, alpha representa el número de éxitos observados (más uno, en una interpretación bayesiana clásica), mientras que beta representa el número de fracasos observados (también más uno). Cuanto mayores sean ambos valores, mayor será la confianza en la estimación y menor será la incertidumbre de la distribución.

Una de las propiedades más importantes de la distribución Beta es que constituye la distribución conjugada de la distribución Bernoulli y Binomial. Esto significa que, al observar nuevos datos, basta con actualizar los parámetros alpha y beta para obtener directamente la nueva distribución posterior, sin necesidad de realizar cálculos complejos. Esta propiedad convierte a la distribución Beta en una herramienta fundamental para la inferencia bayesiana sobre probabilidades desconocidas.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta

alpha = 200
beta_param = 300

x = np.linspace(0, 1, 500)

plt.plot(x, beta.pdf(x, alpha, beta_param))
plt.xlabel("p")
plt.ylabel("densidad")
plt.grid(True)
plt.show()
```
