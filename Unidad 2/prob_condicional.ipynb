```python
import pandas as pd
data=pd.read_excel('edades con escuelas.xlsx')
data.head()
```

```python
hist=data['Edad'].value_counts().sort_index()/len(data)
hist.plot()
```

```python
import plotly.express as px
df=pd.pivot_table(data, index='Edad', columns='Escuela Programa Actual Pregrado', values='Escuela Programa Actual Pregrado', aggfunc='count', fill_value=0).sort_index()
for co in df.columns: 
    df[co]=df[co]/df[co].sum()
df
px.line(df, df.index, df.columns)
```

```python
# Calculemos varias probabilidades para nuestra interpretación: 
p1=len(data[data['Edad']<=24])/len(data)
print('Probabilidad de tener menos de 24 años:',p1)

p2=len(data[(data['Edad']<=24) & (data['Escuela Programa Actual Pregrado']=='Ingeniería-Med')])/len(data)
print('Probabilidad de tener menos de 24 años y ser de ingenierías:',p2)

df2=data[data['Escuela Programa Actual Pregrado']=='Ingeniería-Med']
p3=len(df2[df2['Edad']<=24])/len(df2)
print('Probabilidad de tener menos de 24 años dado que el estudiante es de ingenierías:',p3)
```
